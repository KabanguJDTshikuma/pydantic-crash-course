```python
class User(BaseModel):
    name: str
    email: str

def greet_user(user: User):
    return f"Hello, {user.name}! Your email is {user.email}."

def load_user(data: dict) -> User:
    return User.model_validate(data)
```

```python
user = load_user({"age": 20, "name": "Pat", "email": "testt@test.com"})
message = greet_user(user)
print(message)
```

```python
class WeatherResponse(BaseModel):
    city: str
    temperature: float
    humidity: int
    description: str
```

```python
api_data = {
    "city": "Cape Town",
    "temperature": 22.5,
    "humidity": 60,
    "description": "Partly cloudy"
}
```

```python
weather = WeatherResponse.model_validate(api_data)
```

```python
print(f"Weather in {weather.city}: {weather.description}, Temp: {weather.temperature}°C, Humidity: {weather.humidity}%")
```

```python


from pydantic import Field


class User(BaseModel):
    name: str = Field(min_length=1, max_length=100)
    age: int = Field(gt=0, lt=120)
    email: str
```

```python
user = User(name="Alice", age=23, email="alice@test.com")
print(user)
```

```python
# Nested Models

class OrderItem(BaseModel):
    product_id: int
    quantity: int
    price: float
    name: str

class Order(BaseModel):
    order_id: str
    item: OrderItem
```

```python
order = Order(
    order_id="ORD123",
    item=OrderItem(product_id=1, quantity=2, price=19.99, name="Widget")
)
print(order)
```

```python
print(order.item.name)
```

```python
class Order(BaseModel):
    order_id: str
    customer_email: str
    items: list[OrderItem]

order_data = {
    "order_id": "ORD123",
    "customer_email": "test@test.com",
    "items" :[
        {"product_id": 1, "quantity": 2, "price": 19.99, "name": "Widget"},
        {"product_id": 2, "quantity": 1, "price": 9.99, "name": "Gadget"}
    ],
}

order = Order(**order_data)
```

```python
print(f"order: {order.order_id}")
for item in order.items:
    print(f" - {item.name}: {item.quantity} x ${item.price}")
```

```python
from pydantic_settings import BaseSettings, SettingsConfigDict

class Settings(BaseSettings):
    model_config = SettingsConfigDict(env_file=".env")

    api_key: str
    database_url: str
    debug: bool = False

settings = Settings()
print(f"API Key: {settings.api_key}")
print(f"Database URL: {settings.database_url}")

```
